```python
import gymnasium as gym
from gymnasium import spaces
import numpy as np

class SatelliteMissionEnv(gym.Env):
    def __init__(self, seed=42):
        super().__init__()
        self.observation_space = spaces.Box(
            low=0.0, high=1.0, shape=(8,), dtype=np.float32
        )
        self.action_space = spaces.Discrete(4)
        self.rng = np.random.default_rng(seed)
        self.max_steps = 40

    def _get_obs(self):
        return np.concatenate([
            self.priority,
            self.cloud,
            [self.energy, self.time_left / self.max_steps]
        ]).astype(np.float32)

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        if seed is not None:
            self.rng = np.random.default_rng(seed)

        self.priority = self.rng.uniform(0.2, 1.0, 3)
        self.cloud = self.rng.uniform(0.0, 0.8, 3)
        self.energy = 1.0
        self.time_left = self.max_steps
        self.total_reward = 0.0
        return self._get_obs(), {}

    def step(self, action):
        reward = -0.02

        if action < 3:
            if self.energy < 0.12:
                reward -= 0.25
            else:
                i = action
                reward += self.priority[i] * (1.0 - self.cloud[i])
                reward -= 0.08 * (1.0 + self.cloud[i])
                self.energy -= 0.12
                self.priority[i] *= 0.75
        else:
            self.energy = min(1.0, self.energy + 0.35)
            reward -= 0.06

        self.cloud = np.clip(
            self.cloud + self.rng.normal(0, 0.08, 3), 0, 1
        )
        self.priority = np.clip(
            self.priority + self.rng.normal(0, 0.04, 3), 0, 1
        )

        self.time_left -= 1
        terminated = self.time_left <= 0
        self.total_reward += reward

        return self._get_obs(), float(reward), terminated, False, {
            "energy": self.energy,
            "mission_reward": self.total_reward
        }

env = SatelliteMissionEnv()
obs, info = env.reset(seed=42)

print("Observation shape:", obs.shape)
print("Observation space:", env.observation_space.shape)
print("Valid observation:", env.observation_space.contains(obs))
assert env.observation_space.contains(obs)
```

```python
from stable_baselines3 import PPO
from stable_baselines3.common.env_checker import check_env

check_env(env, warn=True)

model = PPO(
    "MlpPolicy",
    env,
    learning_rate=3e-4,
    n_steps=512,
    batch_size=64,
    gamma=0.99,
    gae_lambda=0.95,
    ent_coef=0.01,
    verbose=1,
    seed=42,
    device="cpu"
)

model.learn(total_timesteps=30_000)
model.save("/content/satellite_ppo")

print("\nPPO training complete.")
```

```python
import gymnasium as gym
from gymnasium import spaces
import numpy as np

class RobustSatelliteMissionEnv(gym.Env):
    def __init__(
        self,
        cloud_min=0.0,
        cloud_max=0.8,
        cloud_noise=0.08,
        seed=42
    ):
        super().__init__()

        self.observation_space = spaces.Box(
            low=0.0, high=1.0, shape=(8,), dtype=np.float32
        )
        self.action_space = spaces.Discrete(4)

        self.cloud_min = cloud_min
        self.cloud_max = cloud_max
        self.cloud_noise = cloud_noise
        self.rng = np.random.default_rng(seed)
        self.max_steps = 40

    def _get_obs(self):
        return np.concatenate([
            self.priority,
            self.cloud,
            [self.energy, self.time_left / self.max_steps]
        ]).astype(np.float32)

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)

        if seed is not None:
            self.rng = np.random.default_rng(seed)

        self.priority = self.rng.uniform(0.2, 1.0, 3)
        self.cloud = self.rng.uniform(
            self.cloud_min, self.cloud_max, 3
        )
        self.energy = 1.0
        self.time_left = self.max_steps
        self.total_reward = 0.0

        return self._get_obs(), {}

    def step(self, action):
        reward = -0.02

        if action < 3:
            if self.energy < 0.12:
                reward -= 0.25
            else:
                i = action
                reward += self.priority[i] * (1.0 - self.cloud[i])
                reward -= 0.08 * (1.0 + self.cloud[i])
                self.energy -= 0.12
                self.priority[i] *= 0.75
        else:
            self.energy = min(1.0, self.energy + 0.35)
            reward -= 0.06

        self.cloud = np.clip(
            self.cloud + self.rng.normal(
                0, self.cloud_noise, 3
            ),
            0, 1
        )
        self.priority = np.clip(
            self.priority + self.rng.normal(0, 0.04, 3),
            0, 1
        )

        self.time_left -= 1
        terminated = self.time_left <= 0
        self.total_reward += reward

        return self._get_obs(), float(reward), terminated, False, {
            "energy": self.energy,
            "mission_reward": self.total_reward
        }

print("Configurable robustness-test environment ready.")
```

```python
import pandas as pd
import numpy as np

def evaluate_robustness(test_env, policy, episodes=100):
    episode_rewards = []

    for ep in range(episodes):
        obs, _ = test_env.reset(seed=5000 + ep)
        total = 0.0
        done = False

        while not done:
            if policy == "PPO":
                action, _ = model.predict(obs, deterministic=True)
                action = int(action)
            else:
                priorities = obs[:3]
                clouds = obs[3:6]
                energy = obs[6]

                if energy < 0.25:
                    action = 3
                else:
                    action = int(np.argmax(priorities * (1 - clouds)))

            obs, reward, terminated, truncated, _ = test_env.step(action)
            total += reward
            done = terminated or truncated

        episode_rewards.append(total)

    return np.mean(episode_rewards), np.std(episode_rewards)

conditions = {
    "Normal clouds": RobustSatelliteMissionEnv(
        cloud_min=0.0, cloud_max=0.8, cloud_noise=0.08
    ),
    "Harsh clouds": RobustSatelliteMissionEnv(
        cloud_min=0.6, cloud_max=1.0, cloud_noise=0.15
    )
}

rows = []

for condition, test_env in conditions.items():
    for policy in ["PPO", "Greedy"]:
        mean_reward, std_reward = evaluate_robustness(
            test_env, policy
        )
        rows.append({
            "Condition": condition,
            "Policy": policy,
            "Mean reward": mean_reward,
            "Std reward": std_reward
        })

robustness_results = pd.DataFrame(rows)
print(robustness_results.round(3).to_string(index=False))
```

```python

def evaluate_mission_metrics(test_env, policy, episodes=100):
    episode_rows = []

    for ep in range(episodes):
        obs, _ = test_env.reset(seed=8000 + ep)

        total_reward = 0.0
        observations = 0
        recharges = 0
        low_energy_actions = 0
        done = False

        while not done:
            if policy == "PPO":
                action, _ = model.predict(obs, deterministic=True)
                action = int(action)
            else:
                priorities = obs[:3]
                clouds = obs[3:6]
                energy = obs[6]

                if energy < 0.25:
                    action = 3
                else:
                    action = int(np.argmax(priorities * (1 - clouds)))

            if action < 3:
                observations += 1
                if obs[6] < 0.12:
                    low_energy_actions += 1
            else:
                recharges += 1

            obs, reward, terminated, truncated, info = test_env.step(action)
            total_reward += reward
            done = terminated or truncated

        episode_rows.append({
            "reward": total_reward,
            "observations": observations,
            "recharges": recharges,
            "low_energy_actions": low_energy_actions,
            "final_energy": info["energy"]
        })

    df = pd.DataFrame(episode_rows)

    return {
        "Mean reward": df["reward"].mean(),
        "Mean observations": df["observations"].mean(),
        "Mean recharges": df["recharges"].mean(),
        "Mean low-energy actions": df["low_energy_actions"].mean(),
        "Mean final energy": df["final_energy"].mean()
    }

metric_rows = []

for condition, test_env in conditions.items():
    for policy in ["PPO", "Greedy"]:
        metrics = evaluate_mission_metrics(test_env, policy)
        metric_rows.append({
            "Condition": condition,
            "Policy": policy,
            **metrics
        })

mission_metrics = pd.DataFrame(metric_rows)

print(mission_metrics.round(3).to_string(index=False))
```

```python

def paired_evaluation(test_env, episodes=200):
    rows = []

    for ep in range(episodes):
        seed = 12000 + ep

        # PPO episode
        obs, _ = test_env.reset(seed=seed)
        ppo_reward = 0.0
        done = False

        while not done:
            action, _ = model.predict(obs, deterministic=True)
            obs, reward, terminated, truncated, _ = test_env.step(int(action))
            ppo_reward += reward
            done = terminated or truncated

        # Greedy episode with the same initial seed
        obs, _ = test_env.reset(seed=seed)
        greedy_reward = 0.0
        done = False

        while not done:
            priorities = obs[:3]
            clouds = obs[3:6]
            energy = obs[6]

            if energy < 0.25:
                action = 3
            else:
                action = int(np.argmax(priorities * (1 - clouds)))

            obs, reward, terminated, truncated, _ = test_env.step(action)
            greedy_reward += reward
            done = terminated or truncated

        rows.append({
            "PPO reward": ppo_reward,
            "Greedy reward": greedy_reward,
            "Difference": ppo_reward - greedy_reward
        })

    df = pd.DataFrame(rows)
    differences = df["Difference"].to_numpy()

    # Approximate 95% confidence interval for the paired mean difference
    mean_diff = differences.mean()
    standard_error = differences.std(ddof=1) / np.sqrt(len(differences))
    ci_low = mean_diff - 1.96 * standard_error
    ci_high = mean_diff + 1.96 * standard_error

    return {
        "PPO mean": df["PPO reward"].mean(),
        "Greedy mean": df["Greedy reward"].mean(),
        "Mean reward advantage": mean_diff,
        "95% CI lower": ci_low,
        "95% CI upper": ci_high
    }

for condition, test_env in conditions.items():
    result = paired_evaluation(test_env)
    print(f"\n{condition}")
    for metric, value in result.items():
        print(f"{metric}: {value:.3f}")
```

```python
import gymnasium as gym
from gymnasium import spaces
import numpy as np

class SatelliteCaptureEnv(gym.Env):
    """
    Synthetic satellite observation scheduling environment.
    This is a research simulator, not a physical flight model.
    """

    def __init__(self, seed=42, cloud_min=0.0, cloud_max=0.8):
        super().__init__()

        self.observation_space = spaces.Box(
            low=0.0, high=1.0, shape=(8,), dtype=np.float32
        )
        self.action_space = spaces.Discrete(4)

        self.initial_seed = seed
        self.cloud_min = cloud_min
        self.cloud_max = cloud_max
        self.rng = np.random.default_rng(seed)
        self.max_steps = 40

    def _get_obs(self):
        return np.concatenate([
            self.priority,
            self.cloud,
            [self.energy, self.time_left / self.max_steps]
        ]).astype(np.float32)

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)

        if seed is not None:
            self.rng = np.random.default_rng(seed)

        self.priority = self.rng.uniform(0.2, 1.0, 3)
        self.cloud = self.rng.uniform(
            self.cloud_min, self.cloud_max, 3
        )
        self.energy = 1.0
        self.time_left = self.max_steps

        self.successful_captures = 0
        self.failed_captures = 0
        self.recharges = 0
        self.mission_reward = 0.0

        return self._get_obs(), {}

    def step(self, action):
        reward = -0.02
        success = False

        if action < 3:
            target = int(action)

            if self.energy < 0.12:
                reward -= 0.25
            else:
                # Every observation attempt consumes energy.
                self.energy -= 0.12

                # Synthetic probability of capture success.
                success_probability = 0.95 * (1.0 - self.cloud[target])
                success = self.rng.random() < success_probability

                if success:
                    reward += float(self.priority[target])
                    self.successful_captures += 1

                    # Reduce priority after a successful capture.
                    self.priority[target] *= 0.60
                else:
                    reward -= 0.10
                    self.failed_captures += 1

        elif action == 3:
            self.energy = min(1.0, self.energy + 0.35)
            self.recharges += 1
            reward -= 0.08

        else:
            raise ValueError("Action must be an integer from 0 to 3.")

        # Weather and target priorities change over time.
        self.cloud = np.clip(
            self.cloud + self.rng.normal(0, 0.08, 3), 0, 1
        )
        self.priority = np.clip(
            self.priority + self.rng.normal(0, 0.04, 3), 0, 1
        )

        self.time_left -= 1
        terminated = self.time_left <= 0
        self.mission_reward += reward

        info = {
            "success": success,
            "successful_captures": self.successful_captures,
            "failed_captures": self.failed_captures,
            "recharges": self.recharges,
            "energy": self.energy,
            "mission_reward": self.mission_reward
        }

        return self._get_obs(), float(reward), terminated, False, info


capture_env = SatelliteCaptureEnv(seed=42)
obs, info = capture_env.reset(seed=42)

print("Observation shape:", obs.shape)
print("Valid observation:", capture_env.observation_space.contains(obs))
print("Available actions:", capture_env.action_space.n)
print("Initial energy:", obs[6])
print("Initial cloud levels:", obs[3:6])

assert capture_env.observation_space.contains(obs)
```

```python

def evaluate_policy(policy_name, episodes=100):
    rewards = []
    successes = []
    failures = []
    recharges = []

    for seed in range(1000, 1000 + episodes):
        env = SatelliteCaptureEnv(seed=seed)
        obs, _ = env.reset(seed=seed)
        done = False
        episode_reward = 0.0

        while not done:
            if policy_name == "PPO":
                action, _ = capture_model.predict(obs, deterministic=True)
                action = int(action)
            elif policy_name == "Random":
                action = env.action_space.sample()
            else:  # Greedy baseline
                energy = obs[6]
                priorities = obs[:3]
                clouds = obs[3:6]

                if energy < 0.24:
                    action = 3
                else:
                    values = priorities * (1 - clouds)
                    action = int(np.argmax(values))

            obs, reward, terminated, truncated, info = env.step(action)
            episode_reward += reward
            done = terminated or truncated

        rewards.append(episode_reward)
        successes.append(info["successful_captures"])
        failures.append(info["failed_captures"])
        recharges.append(info["recharges"])

    return {
        "Policy": policy_name,
        "Mean reward": round(float(np.mean(rewards)), 3),
        "Reward std": round(float(np.std(rewards)), 3),
        "Mean successful captures": round(float(np.mean(successes)), 2),
        "Mean failed captures": round(float(np.mean(failures)), 2),
        "Mean recharges": round(float(np.mean(recharges)), 2),
    }

import pandas as pd

results = pd.DataFrame([
    evaluate_policy("PPO"),
    evaluate_policy("Greedy"),
    evaluate_policy("Random")
])

display(results.sort_values("Mean reward", ascending=False).reset_index(drop=True))
```

```python

def run_episode(policy, seed):
    env = SatelliteCaptureEnv(seed=seed)
    obs, _ = env.reset(seed=seed)
    total_reward = 0.0
    done = False

    while not done:
        if policy == "PPO":
            action, _ = capture_model.predict(
                obs, deterministic=True
            )
            action = int(action)
        else:
            energy = obs[6]
            priorities = obs[:3]
            clouds = obs[3:6]

            if energy < 0.24:
                action = 3
            else:
                action = int(np.argmax(priorities * (1 - clouds)))

        obs, reward, terminated, truncated, _ = env.step(action)
        total_reward += reward
        done = terminated or truncated

    return total_reward

ppo_rewards = []
greedy_rewards = []

for seed in range(2000, 2100):
    ppo_rewards.append(run_episode("PPO", seed))
    greedy_rewards.append(run_episode("Greedy", seed))

ppo_rewards = np.array(ppo_rewards)
greedy_rewards = np.array(greedy_rewards)
differences = ppo_rewards - greedy_rewards

mean_diff = differences.mean()
se = differences.std(ddof=1) / np.sqrt(len(differences))
ci_low = mean_diff - 1.984 * se
ci_high = mean_diff + 1.984 * se

print("PPO mean reward:   ", round(ppo_rewards.mean(), 3))
print("Greedy mean reward:", round(greedy_rewards.mean(), 3))
print("PPO - Greedy:      ", round(mean_diff, 3))
print("95% CI:            ", (round(ci_low, 3), round(ci_high, 3)))
print("PPO wins:          ", int((differences > 0).sum()), "/ 100")
print("Ties:              ", int((differences == 0).sum()))
print("Greedy wins:       ", int((differences < 0).sum()))
```

```python

def evaluate_weather(policy, cloud_min, cloud_max, episodes=100):
    rewards = []
    captures = []
    failures = []

    for seed in range(3000, 3000 + episodes):
        env = SatelliteCaptureEnv(
            seed=seed,
            cloud_min=cloud_min,
            cloud_max=cloud_max
        )
        obs, _ = env.reset(seed=seed)
        total_reward = 0.0
        done = False

        while not done:
            if policy == "PPO":
                action, _ = capture_model.predict(
                    obs, deterministic=True
                )
                action = int(action)
            elif policy == "Greedy":
                if obs[6] < 0.24:
                    action = 3
                else:
                    action = int(np.argmax(obs[:3] * (1 - obs[3:6])))
            else:
                action = env.action_space.sample()

            obs, reward, terminated, truncated, info = env.step(action)
            total_reward += reward
            done = terminated or truncated

        rewards.append(total_reward)
        captures.append(info["successful_captures"])
        failures.append(info["failed_captures"])

    return {
        "Weather": "Normal" if cloud_max < 1.0 else "Harsh",
        "Policy": policy,
        "Mean reward": round(float(np.mean(rewards)), 3),
        "Reward std": round(float(np.std(rewards)), 3),
        "Successful captures": round(float(np.mean(captures)), 2),
        "Failed captures": round(float(np.mean(failures)), 2)
    }

rows = []
for weather, lo, hi in [
    ("Normal", 0.0, 0.8),
    ("Harsh", 0.6, 1.0)
]:
    for policy in ["PPO", "Greedy", "Random"]:
        result = evaluate_weather(policy, lo, hi)
        result["Weather"] = weather
        rows.append(result)

weather_results = pd.DataFrame(rows)
display(weather_results)
```
